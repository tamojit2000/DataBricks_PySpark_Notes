```python
raw_df = (
    spark.readStream
        .format("cloudFiles")
        .option("cloudFiles.format", "csv")
        .option("cloudFiles.schemaLocation","/Volumes/workspace/data/auto_loader_example/schema_location/")
        .option("cloudFiles.schemaEvolutionMode", "addNewColumns")
        .option("header", "true")   # CSV reader option (not cloudFiles.*)
        .load("/Volumes/workspace/data/auto_loader_example/pre_bronze/")
)
```

```python
raw_df.writeStream \
    .format("delta") \
    .option("checkpointLocation","/Volumes/workspace/data/auto_loader_example/checkpoint_location/") \
    .trigger(availableNow=True) \
    .outputMode("append") \
    .start("/Volumes/workspace/data/auto_loader_example/bronze/")
```
