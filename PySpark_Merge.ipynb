```python
df1=spark.createDataFrame(
    [
        ['A',1],
        ['B',2]
    ],
    ['Name','Age']
)
df1.display()
```

```python
df2=spark.createDataFrame(
    [
        ['B',10],
        ['C',100]
    ],
    ['Name','Age']
)
df2.display()
```

```python
from delta.tables import DeltaTable

# write df1 as delta
df1.write.format("delta").mode("overwrite").save("/Volumes/workspace/data/marks/tmp")

# load delta table
delta_df = DeltaTable.forPath(spark, "/Volumes/workspace/data/marks/tmp")

# merge with proper aliases

delta_df.alias("target")\
    .merge(
        df2.alias("source"),
        "target.Name = source.Name"
    )\
    .whenMatchedUpdateAll()\
    .whenNotMatchedInsertAll()\
    .execute()


display(delta_df.toDF())

```
